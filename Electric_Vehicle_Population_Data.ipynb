```python
df = pd.read_csv('Electric_Vehicle_Population_Data.csv')
df.sort_values(by='State')
```

```python
df.groupby('State').agg({'Electric Range': 'mean'}).sort_values('Electric Range', ascending=False).head()
```

```python
df.groupby('State')[['City', 'Make', 'Model']].value_counts()
```

```python
df_WA = df.loc[df['State'] == 'WA']
df_WA.sort_values(by='City', ascending=True)
```

```python
crosstab_city_model = pd.crosstab(df_WA['City'], df_WA['Make'])
crosstab_city_model.loc['Total'] = crosstab_city_model.sum()

def total_cars_for_city(row):
    return row.sum()

crosstab_city_model['Total_cars_for_city'] = crosstab_city_model.apply(total_cars_for_city, axis=1)

crosstab_city_model
```

```python
crosstab_make_model = pd.crosstab(df_WA['Make'], df_WA['Model'])
crosstab_make_model.loc['Total'] = crosstab_make_model.sum()
crosstab_make_model['Total_auto_for_make'] = crosstab_make_model.sum(axis=1)

crosstab_make_model
```
